## SQL Server

```python
rows = list(sales.itertuples(index = False, name = None))
rows
```

```python
import pyodbc
```

```python
conn = pyodbc.connect(                          ## Declare connection
    "driver={ODBC Driver 17 for SQL Server};"
    "server=.\\SQLSERVER;"
    "database=Company_SD;"
    "trusted_connection=yes;"
)

cursor = conn.cursor()
```

```python
cursor.execute('create table sales (order_id int , customer varchar(50), product varchar(20) , quantity int , unit_price int , total_amount int)')
```

```python
cursor.executemany('insert into sales values (?,?,?,?,?,?)', rows)
```

```python
df = pd.read_sql_query('select * from sales', conn)
```
